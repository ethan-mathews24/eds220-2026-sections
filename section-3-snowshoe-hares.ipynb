## 6. Clean values 

```python
# Create a new column called sex_simple where

hares['sex_simple'] = hares['sex'].str.strip().str.lower()

hares['sex_simple'].value_counts(dropna = False)
```

```python
hares_n = hares.replace(to_replace = {'sex_simple' : ['pf', 'f?', '?']}, value = {'sex_simple' : 'm?'})

hares_n['sex_simple'].value_counts()
```

```python
hares.loc[(hares['sex_simple'] == 'f?') | (hares['sex_simple'] == '?') | (hares['sex_simple'] == 'pf'), ['sex_simple']] = 'm?'

hares['sex_simple'].value_counts(dropna = False)
```
