## Archive Exploration 

1. What is this data about?

This data is about the occurrence records and associated climate and environmental data for all arthropod prey of listed species in California drylands.

2. Is this data collected in situ by the authors or is it a synthesis of multiple datasets?

This data is a synthesis of multiple datasets.

3. During what time frame were the observations in the dataset collected?

The data was collected from 1945 - 2022.

4. Does this dataset come with an associated metadata file?

Yes, this dataset comes with an associated metadata file.

5.  Does the dataset contain sensitive data?

This data does not contain sensitive data.

Data Citation:

Rachel King, Jenna Braun, Michael Westphal, & CJ Lortie. (2023). Compiled occurrence records for prey items of listed species found in California drylands with associated environmental data. Knowledge Network for Biocomplexity. doi:10.5063/F1VM49RH.

## 6. Update Column Names

```python
prey = prey.rename(columns = {'institutionCode' : 'institution_code',
                             'datasetKey' : 'dataset_key'})
prey
```

## 7. Subsetting

```python
# Select only the preserved specimens from prey

prey = prey[prey['basisOfRecord'] == 'PRESERVED_SPECIMEN'] 
```

```python
# Select observations between 1992 and 2000.

prey = prey[(prey['year'] >= 1992) & (prey['year'] <= 2000)]
prey
```

```python
# Select all rows for columns species and eventDate

prey_dates = prey.loc[:, ['species', 'eventDate']]
prey_dates
```

```python
# Select observations that are greater than 37° latitude and less than 39° latitude.

prey = prey[(prey['lat'] > 37) & (prey['long'] < 39)]
prey
```
